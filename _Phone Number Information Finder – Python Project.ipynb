# 1. Importing Libraries

```python
import phonenumbers
from phonenumbers import geocoder, carrier, timezone, number_type
import pandas as pd

from datetime import datetime
```

#  2. Creating a Function

```python
def get_phone_details(number):
    try:
        parsed = phonenumbers.parse(number)
#  3. Creating a Details Dictionary
        details = {
            "Entered Number": number,
            "International Format": phonenumbers.format_number(parsed, phonenumbers.PhoneNumberFormat.INTERNATIONAL),
            "National Format": phonenumbers.format_number(parsed, phonenumbers.PhoneNumberFormat.NATIONAL),
            "Country": geocoder.description_for_number(parsed, "en"),
            "Carrier": carrier.name_for_number(parsed, "en"),
            "Timezone": timezone.time_zones_for_number(parsed),
            "Valid Number": phonenumbers.is_valid_number(parsed),
            "Possible Number": phonenumbers.is_possible_number(parsed),
            "Number Type": number_type(parsed).name
        }
# 4. Returning Details        
        return details
#  5. If Number Is Wrong   
    except:
        return {"Error": "Invalid phone number format. Include country code like +91, +1, etc."}
# 6. Main Program Starts

print("📱 Advanced Phone Number Info Finder")
print("-----------------------------------")

number = input("Enter phone number with country code (e.g., +919876543210): ")

details = get_phone_details(number)
print("\n🔍 Phone Number Details:\n")

for k, v in details.items():
    print(f"{k}: {v}")
```

```python
# Save results in CSV
file = "phone_number_details.csv"

df = pd.DataFrame([details])
df.to_csv(file, index=False)

print("\n📁 Saved details to:", file)
```
